```r
#use histogram, plot the distribution of seed rankings
ggplot(wmm_tidy, aes(x = seed)) +
    geom_histogram(binwidth = 1, fill = "pink", color = "white") +
    labs(x = "Seed Ranking",
        y = "Number of Teams",
        title = "Distribution of Seed Rankings")
```

```r
#calculate the proportion of teams in each category
wmm_prop <- wmm_tidy |>
    group_by(final_four) |>
    summarise(count = n()) |>
    mutate(proportion = count / sum(count) * 100)
wmm_prop

#use bar chart, plot the proportions of teams reaching the Final Four
ggplot(wmm_prop, aes(x = final_four, y = proportion)) +
    geom_col(fill = "skyblue") +
    labs(x = "Reached Final Four",
         y = "Percentage(%)",
         title = "Proportion of Teams Reaching the Final Four") +
    scale_x_discrete(labels = c("FALSE" = "Not Reached",
                               "TRUE" = "Reached"))
```

Visualization 1: This bar chart shows the percentage of teams reaching the Final Four for each seed ranking. Teams with lower seed numbers have much higher percentage of reaching Final Four, especially teams ranked as seed 1 or 2. As the number of seed increases, the overall percentage decreases. This suggests that the seed ranking may be an important factor in predicting whether a team reaches the Final Four.

```r
wmm_seed <- wmm_tidy |>
    group_by(seed) |>
    summarise(percentage = mean(final_four) * 100)
wmm_seed

ggplot(wmm_seed, aes(x = seed, y = percentage)) +
    geom_col(fill = "pink") +
    labs(x = "Seed Ranking",
        y = "Teams Reaching Final Four(%)",
        title = "Percentage of Teams Reaching the Final Four by Seed")
```

Visualization 2: This bar chart compares the average conference win percentage of teams that reached the Final Four and those that have not. Teams that reached the Final Four have a higher average conference win percentage, about 89%, while that of teams did not reach the Final Four is approximately 76%. This suggests that conference win percentage may be related to reaching the Final Four and could be useful for predicting whether a team reaches the Final Four.

```r
wmm_conf <- wmm_tidy |>
    group_by(final_four) |>
    summarise(conf_wins_pct_mean = mean(conf_wins_pct, na.rm = TRUE))
wmm_conf

ggplot(wmm_conf, aes(x = final_four, y = conf_wins_pct_mean)) +
    geom_col(fill = "skyblue") +
    labs(x = "Reached Final Four",
        y = "Average Conference Win Percentage(%)",
        title = "Average Conference Win Percentage by Final Four Status") +
    scale_x_discrete(labels = c("FALSE" = "Not Reached",
                               "TRUE" = "Reached"))
```
